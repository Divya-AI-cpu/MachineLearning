```python
student1 = np.array([3,4,5])
student2 = np.array([4,4,4])
```

```python
similarity = np.dot(student1, student2)

print("Similarity Score:", similarity)
```

```python
from numpy.linalg import norm

cosine_similarity = np.dot(student1,student2)/(norm(student1)*norm(student2))

print("Cosine Similarity:", cosine_similarity)
```
